# Scala Ejercicio 03

## Diferencia entre Float y Double

Float - Tiene menor precisión y ocupa 32 bits.

Double - Tiene mayor precisión y ocupa 64 bits.

### Observación

Al guardar el número π en un Double conserva más decimales que Float, ya que tiene una mayor precisión.

Para poder probar el numero de decimales admitidos hay que añadir una **f** al final.

```scala
var piDouble: Double = 3.14159265358979323846264338327
var piFloat: Float = 3.14159265358979323846264338327f

var int: Int = 1
var double: Double = 2.2
var string: String = "String"
```
