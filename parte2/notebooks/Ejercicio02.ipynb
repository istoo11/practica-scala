## Diferencia entre val y var

var - Es una variable, la cual puede variar su valor.
val - Es un valor, el cual es fijo y no pude ser cambiado.

### Observación

Al intentar cambiar un val el cual es fijo, saltara un error ya que dira que el valor ya esta asignado.

```scala
val jugador = "Marta"
var puntuacion = 10

var nuevaPuntuacion = puntuacion + 5
var nuevaSuma = nuevaPuntuacion + 3

jugador = "Sara"
```
