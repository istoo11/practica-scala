# Scala Ejercicio 06

### Explicación

Caso 1 - Cumple que handA es mayor que handB

Caso 2 - Cumple que handB es mayor que handA

Caso 3 - Cumple que ambos son mayor que 21

Caso 4 - No cumple ninguna condicion entonces devuelve el mayor (handB)

Caso 5 - No cumple ninguna condicion entonces devuelve el mayor (handA)


```scala
def bust(handA: Int, handB: Int): Int = {
    if (handA > 21 && handB > 21) 0
    else if (handA > 21) handB
    else if (handB > 21) handA
    else handA.max(handB)
}

bust(26,20)
bust(18,22)
bust(24,25)
bust(17,19)
bust(21,20)
```
