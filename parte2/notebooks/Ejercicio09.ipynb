### Explicación

El contador debe ser mutable, para que de ese modo el contador pueda ser aumentado.

De no poder incrementar el contador se haria un bucle infinito ya que nunca se cumpliria la condicion de salida.

Por ultimo, los elementos que cumplen el modelo imperativo son:

    - var i = 0  -> Por usar una variable mutable var
    - 

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0

while (i < manos.length){
    println(bust(manos(i)))
    i = i + 1
}

def bust(puntuacion: Int): Boolean = {
    puntuacion>21
}
```
