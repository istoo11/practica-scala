# Scala Ejercicio 10

### Explicación

En Scala al usar   **::**   se crea una lista nueva ya que las listas son inmutables y no se pueden modificar.

Por ello para añadir un nuevo elemento se tiene que crear una nueva lista con los datos que habia antes.

```scala
val jugadores = List("Alex", "Chen", "Marta")

val jugadoresNuevos =  "Sindhu" :: jugadores
jugadores.length
jugadoresNuevos.length

val jugadoresReversa = jugadoresNuevos.reverse
```
