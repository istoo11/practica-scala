# Scala Ejercicio 13

## Explicación

El contador lo necesita el while porque es su condicion de salida.

El while es la manera de recorrer que necesita la salida ya que estacambiando el valor de i
en cada iteración.

El contador debe ser mutable, para que de ese modo el contador pueda ser aumentado.

Por ultimo, el modo que mas se aproxima al estilo funcional es el `.foreach`


```scala
val manos = Array(17, 24, 21, 26, 18)
var i = 0

while (i < manos.length){
    println(bust(manos(i)))
    i = i + 1
}

manos.foreach(m => println(bust(m)))

def bust(puntuacion: Int): Boolean = {
    puntuacion>21
}
```
