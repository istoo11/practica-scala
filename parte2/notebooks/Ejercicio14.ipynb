# Scala Ejercicio 14

| Caracteristica | sumarAlTotal | Sumar |
| :--- | :--- | :--- |
| Modifica datos externos | Si | No |
| Devuelve un resultado calculado | No | Si |
| Utiliza efecto secundario | Si | No |
| Estilo predominante | Imperativo | Funcional |

```scala
var total = 0

sumarAlTotal(1)
println(total)
sumarAlTotal(1)
println(total)
sumarAlTotal(1)
println(total)
sumarAlTotal(1)
println(total)

def sumarAlTotal(valor: Int) = {
  total = total + valor
}
```

```scala
def sumar(a: Int, b: Int): Int = {
  a+b
}

var valor1 = 10
var valor2 = 10

println(sumar(valor1,valor2))
```
