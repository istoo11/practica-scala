| Expresión | Resultado | Explicación |
| :--- | :--- | :--- |
| handA > handB | false | handA no es mas grande que handB |
| handB == 21 | true | handB no es igual a 21 |
| handC != 21 | true | handC es diferente a 21 |
| handA <= 21 && handB <= 21 | true | handA $ handB son ambas menores o iguales a 21  |
| handA > 21 \|\| handC > 21 | true | handC es mayor a 21 |
| !(handB>21) | true | handB no es mayor a 21 pero al negarlo devuelva true |



```scala
val handA = 18
val handB = 21
val handC = 25

println(handA > handB)
println(handB == 21)
println(handC != 21)
println(handA <= 21 && handB <= 21)
println(handA > 21 || handC > 21)
println(!(handB>21))
```
