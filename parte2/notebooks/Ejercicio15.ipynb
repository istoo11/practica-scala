## Explicación

- Tanto el contador del bucle, como la lista que guarda los valores son mutables ya que son valores que cambian constantemente,

- Los datos inmutables son los datos de las manos y de los nombres de los jugadores, ya que no van a cambiar nunca.

- Utilizao funciones en las partes que se repiten para asi no duplicar codigo.

- Utilizo estructuras de control en la funcion que comoprueba cual es la mejor puntuacion.

- La parte que se parece mas al estilo imperativo es la funcion preciso ya que controlo yo manualmente el bucle.

- La parte que se parece mas al estilo imperativo es el uso de la funcion bust y el uso del `.foreach` ya que no modifica valores externos, y con los mismos datos de entrada devuelve los mismos datos de salida.

```scala
def bust(puntuacion: Int): Boolean = puntuacion>21
def preciso(puntuaciones: Array[Int]): Int ={
    var filtrados = List[Int]()
    var i = 0
    while(i < puntuaciones.length){
        if(puntuaciones(i)<22){ 
           filtrados = puntuaciones(i) :: filtrados
        }
        i = i + 1
    }
    filtrados.max
}

val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)
manos.foreach(m => println(bust(m)))
println(preciso(manos))

val manosRonda2 = Array(22, 19, 20, 21)
manosRonda2.foreach(m => println(bust(m)))
println(preciso(manosRonda2))
```
