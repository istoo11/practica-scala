# Scala Ejercicio 05

### Observación

Al coincidir ambos numeros, no se cumple ninguna condicion por lo cual la aplicacion no saldria por ningun lado y daría error.

Por ello he añadido una tercera condicion que devuelve 0 en caso de ser iguales.

```scala
def maxHand(a: Int, b: Int): Int = {
    if (a > b) {
        a
    } else if (b > a) {
        b
    } else {
        0
    }
}

maxHand(17, 19)
maxHand(20, 18)
maxHand(21, 21)
```
